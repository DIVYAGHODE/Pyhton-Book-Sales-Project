## **Book Sales Analysis**

```python
# Importing required Python libraries for data analysis and visualization

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the Books Sales dataset into a Pandas DataFrame

df = pd.read_csv("Books_Data_Clean.csv")
```

```python
# Displaying the first five rows of the dataset to understand its structure

df.head()
```

```python
# Generating statistical summary of numerical columns

df.describe()
```

```python
# Filtering the dataset to keep books published after the year 1900

df = df[df["Publishing Year"] > 1900]
```

```python
# Checking the number of missing values in each column

df.isna().sum()
```

```python
# Removing rows where the Book Name is missing

df.dropna(subset = "Book Name", inplace = True)
```

```python
# Checking for duplicate records in the dataset

df.duplicated().sum()
```

```python
# Checking the number of unique values in each column

df.nunique()
```

## **Book Analysis**

```python
# Visualizing the distribution of books according to their publishing year

plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

```python
# Analyzing the number of books available in each genre

df["genre"].value_counts().plot(kind = "bar")
plt.xlabel("Genre")
plt.ylabel("Number of Books")
plt.title("Number of Books in Each Genre")
plt.show()
```

```python
# Calculating the average book rating for each author
# Sorting authors from highest to lowest average rating

df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending = False)
```

```python
# Comparing the distribution of book rating counts across different genres

sns.boxplot(x = "genre", y = "Book_ratings_count", data = df)
plt.xlabel("Genre")
plt.ylabel("Book Rating Count")
plt.title("Box Plot of Book rating Count for Each Genre")
plt.show()
```

## **Sales Analysis**

```python
# Analyzing the relationship between sale price and number of units sold

plt.scatter(df["sale price"], df["units sold"])
plt.xlabel("Sale Price")
plt.ylabel("Units Sold")
plt.title("Scatter Plot of Sale Price vs Units Sold")
plt.show()
```

```python
# Calculating the number of books available in each language

language_counts = df["language_code"].value_counts()
```

```python
# Visualizing the language distribution using a pie chart

plt.pie(language_counts, labels = language_counts.index, startangle = 90, autopct = "%1.1f%%")
plt.title("language Distribution of Books")
plt.show()
```

```python
# Calculating total publisher revenue for each publisher
# Sorting publishers from highest to lowest revenue

df.groupby("Publisher ")["publisher revenue"].sum().sort_values(ascending = False)
```

## **Rating Analysis**

```python
# Finding the maximum average book ratings count by author rating

df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values(ascending = False).max()
```

```python
# Counting the number of books available for each language

df.groupby("language_code").size().sort_values(ascending = False)
```

```python
# Calculating the variance of book rating counts for each author rating category

df.groupby("Author_Rating")["Book_ratings_count"].var()
```

```python
# Analyzing the relationship between average book rating and number of ratings

plt.scatter(df["Book_average_rating"], df["Book_ratings_count"])
plt.xlabel("Book Average Rating")
plt.ylabel("Book Ratings count")
plt.title("Scatter Plot of Book Average Rating vs Book Rating Count")
plt.show()
```

```python
# Calculating total gross sales generated by each author

total_gross_sales_by_author = df.groupby("Author")["gross sales"].sum()
```

```python
# Displaying the top 20 authors based on total gross sales

total_gross_sales_by_author.sort_values(ascending = False).head(20).plot(kind = "bar")
plt.xlabel("Author")
plt.ylabel("Total Gross Sales")
plt.title("Total Gross Sales for Each Author")
plt.show()
```

```python
# Comparing the distribution of units sold across different author rating categories

sns.boxplot(x = "Author_Rating", y = "units sold", data = df)
plt.xlabel("Author Rating")
plt.ylabel("Units Sold")
plt.title("Box Plot of Units Sold for Each Auhtor Rating")
plt.show()
```

```python
# Analyzing the trend of total units sold over the years

df.groupby("Publishing Year")["units sold"].sum().plot(kind = "line", marker = "o")
plt.xlabel("Publishing Year")
plt.ylabel("Total Units Sold")
plt.title("Total Units Sold Over The Years")
plt.show()
```

## **Conclusion**

The analysis shows that Genre Fiction has the highest number of books, while Bill Watterson has the highest average book rating.
English is the most common language, and Penguin Group (USA) LLC generates the highest publisher revenue.
Books with Excellent author ratings show higher units sold, while total units sold show an increasing trend with major fluctuations in recent publishing years.
